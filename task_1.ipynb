```python
from collections import OrderedDict

import numpy as np

from solution_task1 import ShannonFanoEliasBinaryCoder
```

```python
probabilities = OrderedDict(
    [("a", 0.75), ("b", 0.025), ("c", 0.1), ("d", 0.05), ("e", 0.075)]
)
coder = ShannonFanoEliasBinaryCoder(probabilities)
```

```python
print(f"{'Символ':<10} {'Вероятность':<15} {'Код':<15} {'Длина':<10}")
print("-" * 50)

for symbol, prob in probabilities.items():
    code = "".join([c.value for c in coder._code_table[symbol]])
    print(f"{symbol:<10} {prob:<15} {code:<15} {len(code):<10}")


print(f"Энтропия источника: {coder.entropy:.4f}")
print(f"Средняя длина кода: {coder.expected_code_length:.4f}")
print(f"Эффективность кода: {coder.coding_efficiency:.4f}")
```

```python
np.random.seed(42)
symbols = ["a", "b", "c", "d", "e"]
probs = [0.75, 0.025, 0.1, 0.05, 0.075]
message = np.random.choice(symbols, size=1000, p=probs)


encoded_message = []
for symbol in message:
    code_word = coder._code_table[symbol]
    encoded_message.extend(code_word)


print("".join([bit.value for bit in encoded_message[:200]]))

print(f"Длина закодированного сообщения: {len(encoded_message)}")
print(f"Средняя длина кода символа: {len(encoded_message)/1000:.4f}")
print(f"Средняя длина (теоретическая): {coder.expected_code_length:.4f}")
```

### <center> Сравнение с кодом Шеннона и Хаффмана <center>

```python
from shannon import ShannonEncoder
from huffman import BinaryHuffmanEncoder


probabilities = OrderedDict(
    [("a", 0.75), ("b", 0.025), ("c", 0.1), ("d", 0.05), ("e", 0.075)]
)


sfe_coder = ShannonFanoEliasBinaryCoder(probabilities)
shannon_coder = ShannonEncoder(probabilities, ["0", "1"])
huffman_coder = BinaryHuffmanEncoder(probabilities)


print(
    f"{'Кодер':<25} {'Средняя длина':<20} {'Энтропия':<15} {'Эффективность':<15} {'Избыточность':<15} {'Размер закодированного текста':<15}"
)
print("-" * 130)

for name, coder in [
    ("Шеннон-Фано-Элиас", sfe_coder),
    ("Шеннон", shannon_coder),
    ("Хаффман", huffman_coder),
]:
    avg_len = coder.expected_code_length
    entropy = coder.entropy
    efficiency = coder.coding_efficiency
    redundancy = avg_len - entropy

    encoded_message = []
    for symbol in message:
        code_word = coder._code_table[symbol]
        encoded_message.extend(code_word)

    print(
        f"{name:<25} {avg_len:<20.4f} {entropy:<15.4f} {efficiency:<15.4f} {redundancy:<15.4f} {len(encoded_message)}"
    )
```

Вывод: Код Хаффмана наиболее эфективен, чем код Шеннона и Шеннона-Фано-Элиаса
